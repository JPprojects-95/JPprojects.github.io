```python
import sys
import os

sys.path.append(os.path.abspath("../src"))

import numpy as np
import matplotlib.pyplot as plt

from HestonModel import Heston
from BSpricesEU import BS

#### PARAMETERS #####

S0 = 100
r = 0.02
T = 1



      ## important to satisfy feller's condition 2kappa*theta\geq sigma^2
Var0=0.1        ### Initial Variance
kappa=2         ### Mean reversion 
theta=0.1        ### Long run variance
sigma=0.1    ### vol of vol




rho=-0.5

n = 2000        # time steps
N = 100000      # number of paths


K = 99       #Strike


##### SIMULATION of Heston dynamics ######

S,Vol= Heston(S0,Var0,T,r,sigma,kappa,theta,rho,n,N)

Call=np.exp(-r*T)*np.maximum(S[n]-K,0)

sigmaBS=np.sqrt(theta)


###### VISUALIZATION #######

Heston_Callprice = np.mean(Call)
BS_Callprice, BS_Putprice = BS(S0, K, T, r, sigmaBS)         #### simulate BS with sigma0=sqrt(theta)

minVol=np.min(Vol)

print(minVol)

print("Heston Call price", Heston_Callprice)
print("BS Call price",BS_Callprice)
```
